```python
from rdkit import Chem
from rdkit.Chem import Descriptors
import pandas as pd

descriptor_names = [name for name, func in Descriptors._descList]

def calc_descriptors(smiles):
    mol = Chem.MolFromSmiles(smiles)

    if mol is None:
        return [None] * len(descriptor_names)

    return [func(mol) for name, func in Descriptors._descList]

X = pd.DataFrame(
    df["SMILES"].apply(calc_descriptors).tolist(),
    columns=descriptor_names
)

print(X.shape)
```

```python
X = X.fillna(0)

descriptor_df = X.copy()
descriptor_df["Carcinogenicity"] = df["Carcinogenicity"]

print(descriptor_df.shape)
```

```python
descriptor_df.to_csv(
    "Carcinogenicity_RDKit.csv",
    index=False
)
```
