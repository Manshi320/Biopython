```python
import pandas as pd

df = pd.read_csv("AMES_dataset.csv")

print(df.head())
print(df.shape)
print(df.columns)
```

```python
from rdkit import Chem
from rdkit.Chem import Descriptors

print("RDKit is working!")
```

```python
from rdkit import Chem
from rdkit.Chem import Descriptors
import pandas as pd

descriptor_names = [name for name, func in Descriptors._descList]

def calc_descriptors(smiles):
    mol = Chem.MolFromSmiles(smiles)

    if mol is None:
        return [None] * len(descriptor_names)

    return [func(mol) for name, func in Descriptors._descList]

X = pd.DataFrame(
    df["Drug"].apply(calc_descriptors).tolist(),
    columns=descriptor_names
)

y = df["Y"]

print(X.shape)
```

```python
X = X.fillna(0)

descriptor_df = X.copy()
descriptor_df["Y"] = y

descriptor_df.to_csv("AMES_RDKit_Descriptors.csv", index=False)

print(descriptor_df.shape)
```
