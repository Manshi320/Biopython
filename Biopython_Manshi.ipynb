Transcription:
DNA Coding strand
5' ATGCCGTTAGCCGATGC 3'
|||||||||||||||||||||||

3' TACGGCAATCGGCTACG 5'
DNA Template Strand

```python
coding_strand=Seq("ATGCCGTTAGCCGATGC")
template_strand=coding_strand.reverse_complement()
print(template_strand)
```

```python
mRna_sequence=coding_strand.transcribe()
print(mRna_sequence)
```

Biopython also includes a back-transcription method for going from the mRNA to the coding strand of the DNA:

```python
mRna= Seq("C")

coding_strand_seq= mRna.back_transcribe()
print(coding_strand_seq)
```

Translation:

```python
messengerRNA=Seq("AUGCCGUUAGCCGAUGC")
messengerRNA.translate()
```

We can also translate directly from the coding strand DNA sequence.

The translation tables available in Biopython are based on those from the NCBI:
https://www.ncbi.nlm.nih.gov/Taxonomy/Utils/wprintgc.cgi
By default, translation will use the standard genetic code.Suppose we are dealing with a mitochondrial sequence.We need to tell the translation function to use the relevant genetic code instead:


```python
messengerRNA.translate(table="Vertebrate Mitochondrial")
```

```python
coding_strand.translate(table="Vertebrate Mitochondrial")
```

```python
coding_strand.translate(table=2)
```

We may also want to translate the Nt upto the first in frame stop codon, and then stop(as happens in nature).  

```python
messengerRNA.translate(to_stop=True)
```

```python
messengerRNA.translate()
```

```python
coding_strand.translate(table=2,stop_symbol="%")
```
